```python
def ask_area():
    while True:
        text = input("면적(평): ")
        try:
            area = float(text)
        except ValueError:
            print(" → 숫자만 입력하세요. 예 : 18.5")
            continue
        if area <= 0:
            print(" → 0보다 큰 숫자를 입력하세요.")
            continue
        return area
           
```

```python
ask_area()
```
